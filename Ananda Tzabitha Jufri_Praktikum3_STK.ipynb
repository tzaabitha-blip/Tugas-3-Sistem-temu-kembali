```python
def precision_at_k(hasil, relevan, k):
    return sum(1 for d in hasil[:k] if d in relevan) / k

def recall(hasil, relevan):
    cocok = sum(1 for d in hasil if d in relevan)
    return cocok / len(relevan)

def f1(p, r):
    return 0.0 if p+r==0 else 2*p*r/(p+r)

def average_precision(hasil, relevan):
    hit, total = 0, 0.0
    for i, d in enumerate(hasil, start=1):
        if d in relevan:
            hit += 1
            total += hit / i
    return total / len(relevan)

relevan_1 = {2, 3, 5, 9}  # Lagu yang relevan
hasil_1 = [2, 7, 3, 1, 5]  # 5 hasil teratas

relevan_2 = {1, 4, 8}
hasil_2 = [1, 6, 4, 3, 8]

relevan_3 = {6, 7, 10}
hasil_3 = [6, 2, 7, 5, 10]


print("="*60)
print("PRAKTIKUM 3: EVALUASI MESIN PENCARI LAGU")
print("="*60)

# Kueri 1
p1 = precision_at_k(hasil_1, relevan_1, 5)
r1 = recall(hasil_1, relevan_1)
f1_1 = f1(p1, r1)
ap1 = average_precision(hasil_1, relevan_1)

print(f"\nKueri 1: 'cinta patah hati'")
print(f"  Precision@5: {p1:.3f}")
print(f"  Recall: {r1:.3f}")
print(f"  F1-Score: {f1_1:.3f}")
print(f"  AP: {ap1:.3f}")

# Kueri 2
p2 = precision_at_k(hasil_2, relevan_2, 5)
r2 = recall(hasil_2, relevan_2)
f1_2 = f1(p2, r2)
ap2 = average_precision(hasil_2, relevan_2)

print(f"\nKueri 2: 'rindu'")
print(f"  Precision@5: {p2:.3f}")
print(f"  Recall: {r2:.3f}")
print(f"  F1-Score: {f1_2:.3f}")
print(f"  AP: {ap2:.3f}")

# Kueri 3
p3 = precision_at_k(hasil_3, relevan_3, 5)
r3 = recall(hasil_3, relevan_3)
f1_3 = f1(p3, r3)
ap3 = average_precision(hasil_3, relevan_3)

print(f"\nKueri 3: 'bahagia'")
print(f"  Precision@5: {p3:.3f}")
print(f"  Recall: {r3:.3f}")
print(f"  F1-Score: {f1_3:.3f}")
print(f"  AP: {ap3:.3f}")

# MAP
map_score = (ap1 + ap2 + ap3) / 3
print(f"\nMAP: {map_score:.3f}")
```
