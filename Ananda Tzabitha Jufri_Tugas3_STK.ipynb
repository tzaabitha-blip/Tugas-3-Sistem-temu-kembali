```python
print("="*60)
print("TUGAS RUMAH: EVALUASI SISTEM TEMU KEMBALI")
print("="*60)

# ============================================
# 1. 10 JUDUL SKRIPSI
# ============================================

judul = [
    "Analisis penerimaan AI CHATGPT dalam pembelajaran digital menggunakan pendekatan Technology Acceptance Model",
    "Pengembangan media pembelajaran berbasis Augmented Reality untuk pengenalan molekul secara interaktif pada mata pelajaran kimia di MAN 1 Polewali Mandar",
    "Pengaruh pemanfaatan AI generatif terhadap efisiensi waktu dan produktivitas penulis berita real-time di stasiun berita Tribun Timur",
    "Pengaruh intensitas penggunaan media sosial terhadap konsentrasi belajar mahasiswa Pendidikan Teknik Informatika dan Komputer",
    "Pengembangan sistem informasi sekolah berbasis web di UPTD SD Negeri 193 Barru",
    "Rancang bangun sistem monitoring kualitas air secara real-time berbasis Internet of Things (IoT) untuk budidaya ikan dengan antarmuka Blynk",
    "Implementasi jaringan syaraf tiruan untuk penentuan kualitas kentang berdasarkan fitur warna RGB dan tekstur GLCM",
    "Peningkatan kinerja model untuk analisis sentimen kredit pintar menggunakan IndoBERT, SMOTE, dan Stacking Ensemble",
    "Pengembangan sistem informasi bidang ketersediaan dan produksi pangan berbasis web Dinas Ketahanan Pangan Kota Makassar",
    "Pengaruh penggunaan AI generatif model UTAUT terhadap penyelesaian tugas akademik mahasiswa Prodi PTIK UNM"
]

print("\nDAFTAR 10 JUDUL SKRIPSI:")
for i, j in enumerate(judul, 1):
    print(f"{i}. {j}")

# ============================================
# 2. GROUND TRUTH (KUNCI JAWABAN)
# ============================================


relevan_ai = {0, 2, 6, 7, 9}
relevan_jaringan = {1, 6}
relevan_iot = {5}

print("\n" + "="*60)
print("GROUND TRUTH (KUNCI JAWABAN)")
print("="*60)
print(f"Kueri 'AI'     → Relevan: {sorted([d+1 for d in relevan_ai])}")
print(f"Kueri 'Jaringan' → Relevan: {sorted([d+1 for d in relevan_jaringan])}")
print(f"Kueri 'IoT'    → Relevan: {sorted([d+1 for d in relevan_iot])}")

# ============================================
# 3. HASIL PENCARIAN (ASUMSI)
# ============================================

# Hasil pencarian (peringkat dari yang paling mirip)
hasil_ai = [0, 2, 9, 7, 6, 4, 5, 1, 3, 8]
hasil_jaringan = [6, 1, 4, 0, 2, 3, 5, 7, 8, 9]
hasil_iot = [5, 2, 0, 4, 8, 1, 3, 6, 7, 9]

# ============================================
# 4. FUNGSI EVALUASI (DARI MODUL)
# ============================================

def precision_at_k(hasil, relevan, k):
    return sum(1 for d in hasil[:k] if d in relevan) / k

def recall(hasil, relevan):
    cocok = sum(1 for d in hasil if d in relevan)
    return cocok / len(relevan)

def f1(p, r):
    return 0.0 if p+r==0 else 2*p*r/(p+r)

def average_precision(hasil, relevan):
    hit, total = 0, 0.0
    for i, d in enumerate(hasil, start=1):
        if d in relevan:
            hit += 1
            total += hit / i
    return total / len(relevan)

# ============================================
# 5. HITUNG SEMUA METRIK (K=3)
# ============================================

print("\n" + "="*60)
print("HASIL PERHITUNGAN METRIK")
print("="*60)

# --- KUERI 1: AI ---
p1 = precision_at_k(hasil_ai, relevan_ai, 3)
r1 = recall(hasil_ai, relevan_ai)
f1_1 = f1(p1, r1)
ap1 = average_precision(hasil_ai, relevan_ai)

print("\nKUERI: 'AI'")
print(f"  Hasil Top 3: {[d+1 for d in hasil_ai[:3]]}")
print(f"  Precision@3: {p1:.4f}")
print(f"  Recall: {r1:.4f}")
print(f"  F1-Score: {f1_1:.4f}")
print(f"  AP: {ap1:.4f}")

# --- KUERI 2: Jaringan ---
p2 = precision_at_k(hasil_jaringan, relevan_jaringan, 3)
r2 = recall(hasil_jaringan, relevan_jaringan)
f1_2 = f1(p2, r2)
ap2 = average_precision(hasil_jaringan, relevan_jaringan)

print("\nKUERI: 'Jaringan'")
print(f"  Hasil Top 3: {[d+1 for d in hasil_jaringan[:3]]}")
print(f"  Precision@3: {p2:.4f}")
print(f"  Recall: {r2:.4f}")
print(f"  F1-Score: {f1_2:.4f}")
print(f"  AP: {ap2:.4f}")

# --- KUERI 3: IoT ---
p3 = precision_at_k(hasil_iot, relevan_iot, 3)
r3 = recall(hasil_iot, relevan_iot)
f1_3 = f1(p3, r3)
ap3 = average_precision(hasil_iot, relevan_iot)

print("\nKUERI: 'IoT'")
print(f"  Hasil Top 3: {[d+1 for d in hasil_iot[:3]]}")
print(f"  Precision@3: {p3:.4f}")
print(f"  Recall: {r3:.4f}")
print(f"  F1-Score: {f1_3:.4f}")
print(f"  AP: {ap3:.4f}")

# ============================================
# 6. MAP (MEAN AVERAGE PRECISION)
# ============================================

map_score = (ap1 + ap2 + ap3) / 3

print("\n" + "="*60)
print(f"MAP (Mean Average Precision): {map_score:.4f}")
print("="*60)

# ============================================
# 7. TABEL RINGKASAN
# ============================================

print("\n" + "="*60)
print("TABEL RINGKASAN")
print("="*60)
print(f"{'Kueri':<12} {'P@3':<10} {'Recall':<10} {'F1':<10} {'AP':<10}")
print("-"*60)
print(f"{'AI':<12} {p1:<10.4f} {r1:<10.4f} {f1_1:<10.4f} {ap1:<10.4f}")
print(f"{'Jaringan':<12} {p2:<10.4f} {r2:<10.4f} {f1_2:<10.4f} {ap2:<10.4f}")
print(f"{'IoT':<12} {p3:<10.4f} {r3:<10.4f} {f1_3:<10.4f} {ap3:<10.4f}")
print("-"*60)
print(f"{'MAP':<12} {'':<10} {'':<10} {'':<10} {map_score:<10.4f}")
print("="*60)
```
